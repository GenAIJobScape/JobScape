# 3. AI 대체 위협 인식 가설
## 가설 제목
- "AI 기술 발전으로 인한 일자리 대체 우려가 IT 채용 시장을 위축시키고 있다"

## 가설 내용
- 기업들이 AI 도입을 고려하며 신규 개발자 채용을 보류하는 경향
- 특히 반복적 개발 업무(QA, 단순 코딩) 분야의 채용 감소
- 구직자들의 AI 대체 불안감이 취업 준비 방향성에 미치는 영향




- 핵심: AI 위협 인식 ⬆️ ➡️ IT 채용 시장 위축 ⬇️ (특히 특정 직무) & 구직자 전략 변화 ⬆️

- 데이터 연관성 및 시각화 강화 전략 AI 위협 인식:
"한국인 35.4% 'AI가 업무 능력 위협'"은 전반적인 한국인의 인식으로, 이와 함께 'AI 업무 활용자의 본인 업무 대체 가능성 (48.8%)' 데이터를 사용하여 AI를 실제 활용하는 직군(IT/개발자)에서 위협 인식이 훨씬 높음을 강조합니다.

- 구직자 불안감과 행동 변화:
"구직자 73% 'AI 대체 불안감으로 취업 전략 수정 필요'"는 구직자의 행동 변화를 직접적으로 보여줍니다. 여기에 'AI 업무 활용자의 본인/구성원 업무 대체 예상 시기' 데이터를 연결하여, 왜 구직자들이 이러한 불안감을 느끼고 전략을 수정하는지 그 배경(가까운 미래의 대체 가능성)을 설명합니다.

- 특정 IT 직무의 직접적인 영향:
"QA/테스터 채용공고 41% 감소"는 AI의 직접적인 일자리 대체 영향을 보여주는 가장 강력한 증거입니다. 이 데이터는 앞서 AI 위협 인식과 연결되어 **'AI 대체 위협 인식이 높은 직무에서 실제 일자리 감소가 발생하고 있다'**는 흐름을 만듭니다.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as mticker # 퍼센트 포맷팅을 위해 추가

# 한글 폰트 설정 (Colab 환경)
!sudo apt-get install -y fonts-nanum
!sudo fc-cache -fv
!rm ~/.cache/matplotlib -rf
plt.rcParams['font.family'] = 'NanumBarunGothic' # 나눔바른고딕
plt.rcParams['axes.unicode_minus'] = False # 마이너스 기호 깨짐 방지

# --- 기존에 정의된 데이터들 (모두 다시 정의하여 실행 가능하도록 포함) ---
# KISDI 보고서에서 추출한 데이터
# 1. 생성형 AI의 본인 업무 역할 대체 가능성 (N=837)
df_personal_role_replace_total = pd.DataFrame({
    'Category': ['대체할 수 있음', '대체할 수 없음', '보통'],
    'Value': [0.336, 0.269, 0.395],
    'N': 837
})

# 2. 생성형 AI의 본인 업무 역할 대체 예상 시기 (N=281)
df_personal_replace_time_total = pd.DataFrame({
    'Period': ['5년 이내', '6년-10년 이내', '11년-20년 이내', '21년-30년 이내', '31년 이상'],
    'Value': [0.285, 0.463, 0.231, 0.014, 0.007],
    'N': 281
})

# 3. 생성형 AI의 구성원 업무 역할 대체 가능성 (관리직 응답자 N=403)
df_subordinate_role_replace_manager = pd.DataFrame({
    'Category': ['대체할 수 있음', '대체할 수 없음', '보통'],
    'Value': [0.360, 0.275, 0.365],
    'N': 403
})

# 4. 생성형 AI의 구성원 업무 역할 대체 예상 시기 (관리직 응답자 N=145)
df_subordinate_replace_time_manager = pd.DataFrame({
    'Period': ['5년 이내', '6년-10년 이내', '11년-20년 이내', '21년-30년 이내', '31년 이상'],
    'Value': [0.317, 0.462, 0.207, 0.014, 0.000],
    'N': 145
})

# 5. 생성형 AI의 본인 업무 역할 대체 가능성 (업무 활용자 N=289)
df_personal_role_replace_user = pd.DataFrame({
    'Category': ['대체할 수 있음', '대체할 수 없음', '보통'],
    'Value': [0.488, 0.204, 0.308],
    'N': 289
})

# 6. 생성형 AI의 본인 업무 역할 대체 예상 시기 (업무 활용자 N=141)
df_personal_replace_time_user = pd.DataFrame({
    'Period': ['5년 이내', '6년-10년 이내', '11년-20년 이내', '21년-30년 이내', '31년 이상'],
    'Value': [0.277, 0.411, 0.270, 0.035, 0.007],
    'N': 141
})

# 7. 생성형 AI의 구성원 업무 역할 대체 가능성 (업무 활용자 중 관리직 N=165)
df_subordinate_role_replace_user_manager = pd.DataFrame({
    'Category': ['대체할 수 있음', '대체할 수 없음', '보통'],
    'Value': [0.479, 0.218, 0.303],
    'N': 165
})

# 8. 생성형 AI의 구성원 업무 역할 대체 예상 시기 (업무 활용자 중 관리직 N=79)
df_subordinate_replace_time_user_manager = pd.DataFrame({
    'Period': ['5년 이내', '6년-10년 이내', '11년-20년 이내', '21년-30년 이내', '31년 이상'],
    'Value': [0.380, 0.443, 0.152, 0.025, 0.000],
    'N': 79
})

# --- 핵심 요약 데이터 (주어진 3가지 내용) ---
korean_ai_threat_perception = 0.354 # 한국인 35.4% “AI가 업무 능력 위협”
job_seeker_strategy_change = 0.73   # 구직자 73% “AI 대체 불안감으로 취업 전략 수정 필요”
qa_tester_job_decrease = 0.41      # QA/테스터 채용공고 2023-2024년 41% 감소
```

## 시각화 1: AI 업무 활용자(IT/개발자)의 위협 인식과 구직자 불안감
### 내용:
- 전반적인 한국인 인식과 AI를 실제로 다루는 직군(IT/개발자)의 높은 위협 인식을 대조하고, 이것이 구직자들의 행동 변화로 이어진다는 흐름을 보여줍니다.

### 목적:
AI 위협 인식이 단순한 막연한 불안감이 아니라, AI를 직접 다루는 직군에서 더 높게 체감되며, 이것이 구직자들의 취업 전략 수정이라는 구체적인 행동 변화로 이어진다는 것을 강조합니다.

```python
# 시각화 1: AI 업무 활용자(IT/개발자)의 위협 인식과 구직자 불안감

# 데이터 준비
df_ai_threat_perception = pd.DataFrame({
    'Category': ['한국인 AI 업무 위협 인식', 'AI 업무 활용자 역할 대체 가능성 인식', '구직자 취업 전략 수정 필요성'],
    'Value': [korean_ai_threat_perception,
              df_personal_role_replace_user[df_personal_role_replace_user['Category'] == '대체할 수 있음']['Value'].iloc[0],
              job_seeker_strategy_change],
    'Group': ['인식', '인식', '행동 변화']
})

plt.figure(figsize=(10, 7))
sns.barplot(x='Category', y='Value', data=df_ai_threat_perception, hue='Group', palette=['skyblue', 'lightcoral', 'lightgreen'], dodge=False) # dodge=False로 막대 겹치지 않게
plt.title('AI 위협 인식과 구직자 취업 전략 변화의 연관성', fontsize=16)
plt.xlabel('')
plt.ylabel('비율', fontsize=12)
plt.gca().yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1.0))
plt.ylim(0, 0.8)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for index, row in df_ai_threat_perception.iterrows():
    plt.text(row.name, row.Value + 0.02, f'{row.Value:.1%}', color='black', ha="center", va='bottom', fontsize=10)

plt.legend().set_visible(False) # 범례 숨기기
plt.xticks(wrap=True) # 긴 라벨 처리
plt.tight_layout()
plt.show()

print("▶ 시사점: 한국인 전반의 AI 위협 인식(35.4%)을 넘어, AI를 실제로 활용하는 IT/개발자 직군에서는 본인 역할 대체 가능성 인식(48.8%)이 훨씬 높게 나타났습니다. 이러한 위협 인식은 구직자의 73%가 취업 전략 수정을 필요하다고 느끼는 구체적인 행동 변화로 이어지고 있음을 보여줍니다.")
```

## 시각화 2: IT/개발 직군 내 AI 대체 위협 체감 (업무 활용자, 관리직) 및 예상 시기
### 내용:
- IT/개발자 직군의 실무자와 관리자 모두 AI 대체 위협을 체감하고 있음을 보여주며, 특히 구성원 업무의 대체가 더 빠를 것이라는 인식을 통해 기업의 채용 보류 경향을 간접적으로 뒷받침합니다.

###목적:
- AI 위협 인식이 IT/개발 직군 내에서도 보편적이며, 단기적/중기적인 변화가 예상됨을 강조합니다.

```python
# 시각화 2: IT/개발 직군 내 AI 대체 위협 체감 및 예상 시기

# 데이터 준비
df_role_replace_comparison = pd.DataFrame({
    '그룹': ['AI 업무 활용자 (본인)', 'AI 업무 활용 관리직 (구성원)'],
    '대체할 수 있음': [df_personal_role_replace_user[df_personal_role_replace_user['Category'] == '대체할 수 있음']['Value'].iloc[0],
                   df_subordinate_role_replace_user_manager[df_subordinate_role_replace_user_manager['Category'] == '대체할 수 있음']['Value'].iloc[0]],
    '5년 이내 대체 예상': [df_personal_replace_time_user[df_personal_replace_time_user['Period'] == '5년 이내']['Value'].iloc[0],
                        df_subordinate_replace_time_user_manager[df_subordinate_replace_time_user_manager['Period'] == '5년 이내']['Value'].iloc[0]],
    '10년 이내 대체 예상': [df_personal_replace_time_user[df_personal_replace_time_user['Period'].isin(['5년 이내', '6년-10년 이내'])]['Value'].sum(),
                         df_subordinate_replace_time_user_manager[df_subordinate_replace_time_user_manager['Period'].isin(['5년 이내', '6년-10년 이내'])]['Value'].sum()]
})

df_plot_melted = df_role_replace_comparison.melt(id_vars='그룹', var_name='지표', value_name='비율')

plt.figure(figsize=(12, 7))
sns.barplot(x='그룹', y='비율', hue='지표', data=df_plot_melted, palette='coolwarm')
plt.title('AI 업무 활용 직군 (IT/개발자)의 업무 대체 위협 체감', fontsize=16)
plt.xlabel('')
plt.ylabel('비율', fontsize=12)
plt.gca().yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1.0))
plt.ylim(0, 1.0)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for p in plt.gca().patches:
    if p.get_height() > 0:
        plt.gca().annotate(f'{p.get_height():.1%}',
                           (p.get_x() + p.get_width() / 2., p.get_height()),
                           ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontsize=9)

plt.legend(title='인식/예상', loc='upper right', bbox_to_anchor=(1.25, 1))
plt.tight_layout()
plt.show()

print("▶ 시사점: AI를 활용하는 IT/개발 실무자(48.8%)와 관리자(47.9%) 모두 AI가 자신의 업무를 대체할 수 있다고 유사하게 인식합니다. 특히, 관리직은 구성원 업무가 10년 이내에 대체될 가능성(82.3%)을 높게 보고 있어, 기업 내부에서 AI로 인한 인력 구조 개편이 활발하게 논의될 수 있음을 보여줍니다.")
```

## 시각화 3: AI 대체 위협 인식과 실제 IT 직무 채용 감소의 상관 관계

### 내용:
- 높은 AI 대체 위협 인식이 특정 IT 직무의 실제 일자리 감소로 이어지고 있음을 명확히 보여줍니다.

### 목적:
- 추상적인 '인식'이 구체적인 '일자리 감소'로 발현되고 있음을 시각적으로 연결하여 가설의 핵심을 증명합니다.

```python
# 시각화 3: AI 대체 위협 인식과 실제 IT 직무 채용 감소의 상관 관계

# 데이터 준비
df_threat_impact = pd.DataFrame({
    'Indicator': ['한국인 AI 업무 위협 인식', 'AI 업무 활용자 역할 대체 인식', 'QA/테스터 채용 공고 감소'],
    'Value': [korean_ai_threat_perception,
              df_personal_role_replace_user[df_personal_role_replace_user['Category'] == '대체할 수 있음']['Value'].iloc[0],
              qa_tester_job_decrease],
    'Type': ['인식', '인식', '실제 영향']
})

plt.figure(figsize=(10, 7))
sns.barplot(x='Indicator', y='Value', data=df_threat_impact, hue='Type', palette=['lightgray', 'skyblue', 'salmon'], dodge=False)
plt.title('AI 대체 위협 인식과 특정 IT 직무 채용 감소', fontsize=16)
plt.xlabel('')
plt.ylabel('비율', fontsize=12)
plt.gca().yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1.0))
plt.ylim(0, 0.6) # Y축 범위 조정
plt.grid(axis='y', linestyle='--', alpha=0.7)

for index, row in df_threat_impact.iterrows():
    plt.text(row.name, row.Value + 0.02, f'{row.Value:.1%}', color='black', ha="center", va='bottom', fontsize=10)

plt.legend().set_visible(False) # 범례 숨기기
plt.xticks(wrap=True)
plt.tight_layout()
plt.show()

print("▶ 시사점: AI가 업무 능력을 위협한다는 인식(한국인 35.4%, AI 활용자 48.8%)이 높은 가운데, AI로 인해 자동화 가능성이 높은 QA/테스터 직무의 채용 공고가 41%나 급감했습니다. 이는 AI 대체 위협 인식이 단순히 심리적인 요인을 넘어, 실제 IT 채용 시장, 특히 반복적/단순 업무 직무에 직접적인 위축 효과를 미치고 있음을 명확하게 보여줍니다.")
```

### 종합 분석 및 결론 도출

1. **AI 대체 위협 인식의 확산**  
    - 전체 한국인의 35.4%가 AI가 자신의 업무 능력을 위협한다고 인식하고 있으며, 실제 IT/개발 직군(특히 AI를 직접 활용하는 업무자)에서는 이 비율이 48.8%로 더욱 높게 나타남.
    - 관리자 역시 구성원 업무의 AI 대체 가능성을 높게 평가(47.9%)하며, 10년 이내 대체될 것이라는 전망이 우세함.

2. **구직자 행동 변화**  
    - AI 대체 불안감으로 인해 구직자의 73%가 취업 전략을 수정해야 한다고 응답, 이는 단순한 인식 수준을 넘어 실제 행동 변화로 이어지고 있음.

3. **실제 IT 직무 채용 감소**  
    - 반복적/단순 업무(예: QA/테스터) 분야에서 채용공고가 1년 새 41% 감소하는 등, AI 대체 위협 인식이 실제 IT 채용 시장의 위축으로 연결되고 있음.

4. **시사점**  
    - AI 기술 발전은 IT/개발 직군에서 일자리 대체에 대한 불안과 실제 채용 감소라는 이중적 영향을 미치고 있음.
    - 기업은 AI 도입에 따라 인력 구조 재편을 적극적으로 고려하고 있으며, 구직자 역시 이에 맞춰 역량 개발 및 전략 수정이 필수적임.
    - 반복적·단순 업무 중심의 IT 직무는 AI 대체 위험이 높으므로, 향후 IT 인재는 창의적 문제 해결, AI 활용 역량 등 차별화된 경쟁력을 갖추는 것이 중요함.

**결론:**  
AI 대체 위협 인식은 IT 채용 시장의 위축과 구직자 행동 변화, 실제 일자리 감소로 이어지고 있다. IT/개발 직군은 AI 시대에 맞는 역량 강화와 전략적 대응이 필수적이다.
